# Agents

### Agents in essence are LLMs that accomplish a specific task. They can be supplimented with tools, structured output and handoff to other agents

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.environ.get("LLM_API_KEY")

if not api_key:
    raise ValueError("LLM_API_KEY is not set in the environment variables")
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key,
)

completion = client.chat.completions.create(
    model="openai/gpt-5.4-mini",
     max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is the meaning of life?"
        }
    ]
)

print(completion.choices[0].message.content)
```

```python
from openai import AsyncOpenAI

from agents import (
    Agent,
    Runner,
    OpenAIChatCompletionsModel,
    set_tracing_disabled,
)

import os

client = AsyncOpenAI(
    api_key=os.getenv("LLM_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)

set_tracing_disabled(True)

model = OpenAIChatCompletionsModel(
    model="openrouter/free",
    # model="gpt-4o-mini",
    openai_client=client,
)
```

```python
from agents import Agent, Runner

try:

    agent = Agent(
            name = "JARVIS",
            instructions = "You are a personal, resourceful, responsible, helpful and capable assistant.",
            model = model
    )

    result = await Runner.run(agent, "JARVIS! Daddy's Home")
    print(result.final_output)

except Exception as e:
    print("An error occurred:", e)
```

```python
facts_agent = Agent(
        name = "JARVIS",
        instructions = "You are a knowledgeable AI Agent and you keep me updated by general knowledge, current affairs and interesting facts including science, maths, space, computer science, geology, astronomy, physics, history of physicists",
        model = model
)

prompt = "Give me 5 facts"
fact_result = await Runner.run(
                facts_agent,
                prompt
)

print(fact_result.final_output)
```
